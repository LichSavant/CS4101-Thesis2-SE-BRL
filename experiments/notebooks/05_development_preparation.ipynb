```
from pathlib import Path
import sys

cwd = Path.cwd().resolve()
REPO_ROOT = next((p for p in (cwd, *cwd.parents)
                  if (p / "ml" / "notebook_contracts.py").is_file()
                  and (p / "experiments").is_dir()), None)
if REPO_ROOT is None:
    raise RuntimeError("Repository root not found")
if str(REPO_ROOT) not in sys.path:
    sys.path.insert(0, str(REPO_ROOT))
from ml.notebook_contracts import PREPARATION_RUN_ID
from ml.development_experiments import (
    FeatureColumn, DevelopmentSpecification, load_development_dataset,
    development_folds, preparation_summary, CLASSIFIER_FAMILIES,
)
```

## Explicit feature and protocol specification

Populate these from the actual frozen study records. A reference string is
traceability, not proof of approval. Do not infer features from numeric dtypes.
Represent each conventional field as `FeatureColumn(name, "text" or "numeric")`.
The appropriate text vectorizers, numeric transforms and candidate settings must

```
RUN_ID = PREPARATION_RUN_ID
DATASET = "Email"
PROTOCOL_REFERENCE = None
FEATURE_DICTIONARY_REFERENCE = None
CONVENTIONAL_FEATURES = None

if any(value is None for value in (PROTOCOL_REFERENCE, FEATURE_DICTIONARY_REFERENCE, CONVENTIONAL_FEATURES)):
    raise RuntimeError("Provide the frozen protocol and explicit conventional feature dictionary before consuming data")
specification = DevelopmentSpecification(DATASET, CONVENTIONAL_FEATURES,
    PROTOCOL_REFERENCE, FEATURE_DICTIONARY_REFERENCE)
```

```
dataset = load_development_dataset(REPO_ROOT, RUN_ID, specification=specification)
dict(preparation_summary(dataset))
```

```
fold_inputs = {configuration: development_folds(dataset, configuration)
               for configuration in dataset.configurations}
# Counts and identities only; no raw messages, predictions or training outputs.
for configuration, plans in fold_inputs.items():
    print(configuration, [(p.validation_fold, len(p.training_ids), len(p.validation_ids)) for p in plans])
print("Candidate families prescribed by the paper:", CLASSIFIER_FAMILIES)
print("Model fits performed: 0")
```
